```python
from IPython.display import display
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
df = pd.read_csv('respondents.csv', low_memory=False)

print(f'Liczba respondentów (wierszy): {df.shape[0]:,}')
print(f'Liczba pytań/zmiennych (kolumn): {df.shape[1]}')
```

```python
overview = pd.DataFrame({
    'Typ danych': df.dtypes,
    'Niepuste wiersze': df.notna().sum(),
    'Braki (%)': (df.isna().mean() * 100).round(1),
})

display(overview)
display(df.describe().T)
```
